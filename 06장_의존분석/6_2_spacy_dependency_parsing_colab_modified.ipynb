## 3. 패키지 불러오기

원본 예제와 동일한 단계입니다.

```python
import sys
import spacy
from spacy import displacy

print("Python:", sys.version.split()[0])
print("spaCy:", spacy.__version__)
```

## 4. 영어 모델 불러오기

```python
nlp = spacy.load("en_core_web_sm")
print("모델 로드 완료")
```

## 5. 분석 문장 지정

원본 예제의 문장을 그대로 사용합니다.

```python
text = "The fat cat sat on the mat"
print(text)
```

## 6. 문장 분석

```python
doc = nlp(text)
print(doc)
```

## 7. 토큰 출력

```python
for token in doc:
    print(token.text)
```

## 8. 토큰의 품사와 표제어 출력

```python
for token in doc:
    print(
        token.text,
        token.lemma_,
        token.pos_,
        token.tag_,
        token.is_stop
    )
```

## 9. 의존 관계 출력

`token.dep_`는 의존 관계, `token.head.text`는 해당 토큰의 중심어입니다.

```python
for token in doc:
    print(token.text, token.dep_, token.head.text)
```

## 10. 분석 결과를 표로 정리

```python
import pandas as pd

rows = []
for token in doc:
    rows.append({
        "토큰": token.text,
        "표제어": token.lemma_,
        "품사": token.pos_,
        "세부품사": token.tag_,
        "의존관계": token.dep_,
        "중심어": token.head.text,
        "불용어": token.is_stop
    })

df = pd.DataFrame(rows)
df
```

## 11. 의존 구문 분석 시각화

```python
displacy.render(doc, style="dep", jupyter=True)
```

## 12. 시각화 옵션 적용

```python
options = {"compact": True, "distance": 100}

displacy.render(
    doc,
    style="dep",
    jupyter=True,
    options=options
)
```

## 13. 모델 및 파이프라인 확인

```python
print(nlp.pipe_names)
print("문장 수:", len(list(doc.sents)))
```
