```python
# NLTK 라이브러리를 임포트하고 필요한 데이터셋을 다운로드합니다.
import nltk

nltk.download('punkt')
nltk.download('words')
nltk.download('averaged_perceptron_tagger')
nltk.download('maxent_ne_chunker')
nltk.download('punkt_tab')
```

```python
# 분석할 문장을 정의합니다.
sentence = "Prime Minister Boris Johnson had previously said the UK would leave by 31 October."
```

```python
# 문장을 단어 단위로 토큰화합니다.
import nltk
tokens = nltk.word_tokenize(sentence)
print(tokens)
```

```python
# 토큰화된 단어들에 품사(Part-of-Speech) 태그를 지정합니다.
nltk.download('averaged_perceptron_tagger_eng')
tagged = nltk.pos_tag(tokens)
print(tagged)
```

```python
# 품사 태그가 지정된 단어들에서 개체명(Named Entity)을 추출합니다.
nltk.download('maxent_ne_chunker_tab')
entities = nltk.chunk.ne_chunk(tagged)
print(entities)
```
